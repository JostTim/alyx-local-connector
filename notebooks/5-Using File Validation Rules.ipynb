```python
from alyx_connector.files.validation.file_implementation import FileRecordList, FileValidator
```

```python
rule_file = {"rules": {
        "imaging_data_rule1": {
            "if": {
                "extension": {
                    "in": [
                        "tif",
                        "tiff"
                    ]
                },
                "collection": {
                    "contains": "imaging_data"
                }
            },
            "on": {
                "match": {"rename": 
                    {
                    "object": "imaging",
                    "attribute": "frames",
                    "extra": {
                        "pattern": "TIFF_FRAME_NO",
                        "eval": "match[1].zfill(5)"
                        }
                    }
                },
                "destination_exists": "abort",
                "rename_unchanged": "exclude"
            },
        },
    },
    "re_patterns":{
            "TIFF_FRAME_NO": ".*(\\d{5})\\.tiff?$",
    }
}

validator = FileValidator.parse_from_dict(
    rule_file)
```

```python
validator = FileValidator.from_yaml_file(r"rules.yaml")
```

```python
records = FileRecordList.from_iterable([
    r"//prout/wm12/2026-03-12/001/imaging_data/test00001.tiff",
    ])
records
```
