```python
from alyx_connector import Connector
connector = Connector()
```

# List the available endpoints

```python
connector.web_client.list_endpoints()
```

# Use one endpoint to create an object
(taking as an example a mouse)

```python
labs = connector.web_client.rest("labs", "list")
labs
```

```python
lab = connector.web_client.rest("labs", "create", data={"name":"TestingGrounds"})
lab
```

```python
value = connector.web_client.rest("subjects", "create", data={"nickname":"mickey_17", "lab":"TestingGrounds"})
value
```

```python
subjects = connector.web_client.rest("subjects", "list", lab="TestingGrounds")
subjects
```

```python
# A subject's name (unique) is created from lab_name__subject_nickname
# Users can still search by nickname if they want
value = connector.web_client.rest("subjects", "list", nickname="mickey_17")
value
```

```python
# A subject's name (unique) is created from lab_name__subject_nickname
# Users can still search by nickname if they want
value = connector.web_client.rest("subjects", "destroy", name="TestingGrounds__mickey_17")
```
