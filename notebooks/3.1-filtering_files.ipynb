```python
from alyx_connector import Connector
from alyx_connector.files import File, find_files
from alyx_connector.configuration import LocalData
connector = Connector()
```

```python
sessions = connector.search(subject="ea04")
session = sessions.iloc[-1] # here we obtain a session's path
```

```python
from pathlib import Path
relative_path = Path(session.path).relative_to(session.default_data_repository_path)
path = LocalData.directory / relative_path
```

```python
from pandas import Series
from typing import List, Optional

class FileList(List[File]):

    def __init__(self, files: Optional[List[File]] = None, session: Optional[Series] = None):
        super().__init__(files if files else [])
        self.session = session

    
```

```python
files =  FileList([File(file) for file in find_files(path, get="files")])
```
