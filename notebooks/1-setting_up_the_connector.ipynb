## Import the connector class from alyx_connector

```python
from alyx_connector import Connector
```

## For a first user connection, using the .setup() method is recommended
It provides several options to tune which server and user you want to setup, as well as 
assigning them as defaults or not, setting the user config as silent (no prompts), providing a 
password programatically (without a prompt) if needed, etc. By default, if you run .setup() without
arguments, all the values will be asked to you via prompts that won't leave a trace in the notebook or logs,
and save the sensitive infos (username/tokens) only in the alyx-connector configuration files.

In case you want to setup the password via code, witout requireing a user prompt input,
make sure that python obtains the value for the password from environment or config files,
and ***never*** hardcode a password as a plain string, as it is prone to security issues and 
password leak (if the code is shared without noticing the hardcoded password in it, for example).

```python
connector = Connector.setup(url="http://127.0.0.1:80", username="admin")
```

## For subsequent connections to the same config
Works after kernel stop, or using a different instance of python, as the default user and server
is stored in the config

```python
connector = Connector()
```

## Test wether you are logged in or not
It simple checks if a token is present for the server / user, 
not if that server is valid, until any request is made.

```python
connector.web_client.is_logged_in()
```

## Log out a user
It is not necessary to perform this after a session as the tokens stay valid for months, for a given user.
Currentely, logging out only removes the user token from the config, but doesn't invalidate it server wise.

```python
connector.web_client.logout()
```

## Deleting the whole config
If you want to restart fresh after adding too many servers and users, for testing, etc.

```python
connector.config.delete_all()
```
